```python
# Regression of one time series (ts) onto a field with lead lag years. field leads ts
# by '-ly' years. And do block bootstrap significant test 
# the output is for one lead-lag year: lead year '-ly'

length=6 # -5,....0 year
ly = 0
variable = 't_ver'
block_length=2
n_resamples=1000
confidence_level1=0.9
confidence_level2=0.95
bp='bp310'

import pandas as pd
import xarray as xr
import numpy as np

# read data
data       = xr.open_dataset("/work/uo1075/u241321/data/u241321/data_cdo/temperature_1969-2019_assi_dt_5000.nc",decode_times=False)
ts = np.load("/work/uo1075/u241321/data/y310_T.npy")   # T from band-pass filtering, 3-30yr  1972-2017, cut 4 years due to filtering

corr      = xr.open_dataset("/work/uo1075/u241321/data/reg_"+variable+"_tC2_"+bp+".nc")['__xarray_dataarray_variable__'] 


nyear=ts.size
var = np.mean(data['__xarray_dataarray_variable__'], axis=1)[:,0:21,:,:]
var_1 = var.stack(spatial=('depth','lat','lon')).dropna(dim="spatial") #time,space
spatial = var_1.coords["spatial"]

y1 = var_1[length+1+ly:nyear+ly-3]
y2 = ts[length+1:nyear-3]

coef = corr[5+ly,:,:,:].stack(spatial=('depth','lat','lon')).dropna(dim="spatial") #time,space

# regression, center on 6-45 (1976-2015), 40 year (start from 0)

# for m in range(0,field.shape[1],1):
#         coe[0,m] = regression(y[7:47], field[2:42,m])
#         coe[1,m] = regression(y[7:47], field[3:43,m])
#         coe[2,m] = regression(y[7:47], field[4:44,m])
#         coe[3,m] = regression(y[7:47], field[5:45,m])
#         coe[4,m] = regression(y[7:47], field[6:46,m])
#         coe[5,m] = regression(y[7:47], field[7:47,m])


#  LinearRegression

from sklearn.linear_model import LinearRegression
def regression(x,y):

    coef = LinearRegression(fit_intercept=True).fit(x.reshape(-1, 1), y.values.reshape(-1, 1)).coef_[0,0]
    

    return coef


# bootstrap

def _bootstrap_resample(sample1, sample2, block_length, n_resamples=None, random_state=None):
    """Bootstrap resample the sample."""
   
    np.random.shuffle(sample1.reshape((-1,block_length)))
    rng_state = np.random.get_state()
    np.random.set_state(rng_state)
    np.random.shuffle(sample2.values.reshape((-1,block_length)))
    
    # resamples = sample
    coef = regression(sample1, sample2)
    # bootstrap - each row is a random resample of original observations
    # i = rng_integers(random_state, 0, n, (n_resamples, n))

    # resamples = sample[..., i]
    return coef


def bootstrap(data1, data2, rawcoef, block_length, confidence_level1, confidence_level2, n_resamples,
              method, random_state=None):
 
    # the 1st dimension of data is time
    # confidence_level: 0.9, 0.95, 0.99...
    # n_resamples: 1000, 10000...
    # method: 'percentile', 'basic', 'bca'
    
    # the original Pearson correlation coefficients 
    coef_raw = rawcoef
 
    theta_hat_b = []

    batch_nominal = n_resamples

    for k in range(0, n_resamples):
        # batch_actual = min(batch_nominal, n_resamples-k)
        # Generate resamples
        resampled_coef_data = []
        # for sample in data:
        resample = _bootstrap_resample(data1, data2, block_length, 
                                           random_state=random_state)
        resampled_coef_data.append(resample)

        # Compute bootstrap distribution of Pearson correlation coefficients
        theta_hat_b.append(resampled_coef_data)
    theta_hat_b = np.concatenate(theta_hat_b, axis=-1)

    # Calculate percentile interval
    alpha1 = (1 - confidence_level1)/2
    alpha2 = (1 - confidence_level2)/2

    interval1 = alpha1, 1-alpha1
    interval2 = alpha2, 1-alpha2

    def percentile_fun(a, q):
        return np.percentile(a=a, q=q, axis=-1)

    # Calculate confidence interval of statistic
    ci_l1 = percentile_fun(theta_hat_b, interval1[0]*100)
    ci_u1 = percentile_fun(theta_hat_b, interval1[1]*100)
    
    # Calculate confidence interval of statistic
    ci_l2 = percentile_fun(theta_hat_b, interval2[0]*100)
    ci_u2 = percentile_fun(theta_hat_b, interval2[1]*100)
    
      
    
    if coef_raw >= ci_u1:
        if coef_raw >= ci_u2:
            sig = 0.95
        else:
            sig = 0.9
        
    else:
        if coef_raw <= ci_l1:
            if coef_raw <= ci_l2:
                sig = -0.95
            else:
                sig = -0.9
        else:
            sig = 0
        
    return coef_raw ,abs(sig)

# apply to global filed

from multiprocessing import Pool
from itertools import product
from itertools import starmap

def boots_gridpoint(m):
    
    mode = bootstrap(
         y2, y1[:,m], coef[m], block_length=block_length, confidence_level1=confidence_level1, confidence_level2=confidence_level2, 
        n_resamples=n_resamples, method='percentile'
    )

    return mode



# pool = Pool()
# ! python --version

res = Pool().map(boots_gridpoint,np.arange(0,y1.shape[1],1))
re = np.array(res)

# output 
corr = xr.DataArray(re[:,0], dims = ["spatial"], coords = {"spatial":spatial}).unstack()    
sig = xr.DataArray(re[:,1], dims = ["spatial"], coords = {"spatial":spatial}).unstack()    

lat = data['lat']
lon = data['lon']
depth = var.depth
nlat = lat.size
nlon = lon.size

boot = xr.Dataset(data_vars={'lat': (["lat"], lat.data),
                              'lon':  (["lon"], lon.data),
                              'depth':(["depth"], depth.data),
                              'sig': (["depth","lat", "lon"], sig.values),
                              'corr': (["depth","lat", "lon"], corr.values),})
# corr = correlation["corr"]
boot.to_netcdf("/work/uo1075/u241321/data/boot_reg_"+variable+"_ly"+str(-ly)+"_block"+str(block_length)+bp+".nc") 
```
